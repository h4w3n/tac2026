## Utilisation de l'API GATE Cloud API
French Named Entity Recognizer 

```python
# texte sélectionner dans le corpus CAMILLE

camille_text = "Mariages : Gbys, colonel en retraite, à Molenbeek- St-Jean, ef Pien, propriétaire, rue Américaine. — Pe­tit, cocher, à Bruxelles, et Ubioulle, femme déchu bre, rue Eeyenveld. — Deblon, ingénienr, rue l'Arbre-lénit, m de Meaten de Borne, rentière, à Isfiarétiûû*."
```

```python
response = requests.post('https://cloud-api.gate.ac.uk/process-document/french-named-entity-recognizer',
                         auth = ('gchp8zffo8ye', '9334mbox3q1exo2o9j00'),                                   # key ID et password de GATE Cloud
                         params = {'annotations':[':Person', ':Location', ':Sentence', ':Organization']},   # catégorie de POS Tagging 
                         data = camille_text.encode('UTF-8'),                                               # texte à analyser
                         headers ={'Content-Type':'text/plain'})

j = response.json()
txt = j['text']
for ann_type, anns in j['entities'].items():
  print(ann_type)
  for a in anns:
    start, end = a['indices']
    print(f'- {txt[start:end]} ({start}:{end})')

for (k, v) in response.headers.items():
  print(f"{k}: {v}")
```
